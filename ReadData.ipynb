```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('student_grade_ses_dataset.csv')
df.info()
plt.figure(figsize=(10,5))
plt.hist(df[['family_size']], color='pink', edgecolor='black', bins=7)
plt.title('Размер семьи')
plt.xlabel('Значение')
plt.ylabel('Частота')
plt.grid(True, axis = 'y')
plt.show()



```
