```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('Student_Depression_Dataset.csv')
df[['CGPA', 'Depression']].info()
df.boxplot(column='CGPA', by='Depression', figsize=(8,6))
# df сам знает метод boxplot, но чтобы его исполнить надо всё равно импортировать matplotlib(он как маляры)
# так делаем, потому что у plt.matplotlib нет column и by
plt.title('Распределение CGPA у студентов с депрессией и без')
plt.grid(True)
plt.show()

```
