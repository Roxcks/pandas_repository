```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('Student_Depression_Dataset.csv')
df.info()
plt.figure()
plt.boxplot(df[['Cost_without_study', 'CGPA']],tick_labels=['Cost_without_study', 'CGPA'])
plt.title('BoxPlot')
plt.ylabel('Cost')
plt.grid(True)
plt.show()

```
