```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('Student_Depression_Dataset.csv')
df.info()
plt.figure()
plt.scatter(df['Cost_without_study'], df['CGPA'], alpha=0.3)
plt.title('Scatterplot(диаграмма рассеивания)')
plt.xlabel('Cost_without_study')
plt.ylabel('CGPA')
plt.grid(True)
plt.savefig('cost_and_CGPA.png')
plt.show()

```
