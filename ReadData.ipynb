```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('student_grade_ses_dataset.csv')
df.info()
plt.figure(figsize=(10,5))
df_bar = df.groupby('school_type')['study_hours_per_week'].mean()
# связь между двумя критериями + считаем среднее значение для school_type критерия study
plt.bar(df_bar.index, df_bar.values, color='yellow', alpha=0.8)
# индекс - подписи по оси Х (названия категорий school_type), value - высота столбиков исходя из значений
plt.title('Barplot example')
plt.xlabel('Категории')
plt.ylabel('Часы учёбы')
plt.grid(True, axis = 'y')
plt.show()
print(df_bar)



```
